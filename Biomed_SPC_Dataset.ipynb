```python
import os
from collections import Counter

dataset_path = "/content/Dataset"

all_files = []
extensions = Counter()
subjects = {}

for subject in sorted(os.listdir(dataset_path)):
    subject_path = os.path.join(dataset_path, subject)

    if os.path.isdir(subject_path):
        files = []

        for root, dirs, filenames in os.walk(subject_path):
            for filename in filenames:
                full_path = os.path.join(root, filename)
                files.append(full_path)
                extensions[os.path.splitext(filename)[1].lower()] += 1

        subjects[subject] = len(files)
        all_files.extend(files)

print("Total subjects:", len(subjects))
print("Total files:", len(all_files))
print("\nFiles per subject:")

for subject, count in subjects.items():
    print(subject, ":", count)

print("\nFile types:")
for ext, count in extensions.items():
    print(ext or "[no extension]", ":", count)

print("\nTotal size (MB):",
      round(sum(os.path.getsize(f) for f in all_files) / (1024**2), 2))
```

```python
import os
import csv
import hashlib
import zipfile

paths = [
    "/content/Dataset",
    "/content/Biomed_SPC Measuring SMR during Active and Resisti",
    "/content/dataset.zip"
]

manifest_path = "/content/dataset_sha256_manifest.csv"
records = []

# Generate SHA-256 hashes
for base in paths:
    if os.path.isfile(base):
        files = [base]
    elif os.path.isdir(base):
        files = [
            os.path.join(root, name)
            for root, _, names in os.walk(base)
            for name in names
        ]
    else:
        continue

    for path in files:
        digest = hashlib.sha256()

        with open(path, "rb") as f:
            for chunk in iter(lambda: f.read(1024 * 1024), b""):
                digest.update(chunk)

        records.append({
            "file": path,
            "sha256": digest.hexdigest(),
            "size_bytes": os.path.getsize(path)
        })

with open(manifest_path, "w", newline="") as f:
    writer = csv.DictWriter(
        f, fieldnames=["file", "sha256", "size_bytes"]
    )
    writer.writeheader()
    writer.writerows(records)

# Verify against the generated manifest
verified = 0
failed = []

for record in records:
    digest = hashlib.sha256()

    with open(record["file"], "rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            digest.update(chunk)

    if digest.hexdigest() == record["sha256"]:
        verified += 1
    else:
        failed.append(record["file"])

print("Files included:", len(records))
print("Verified:", verified)
print("Failed:", len(failed))
print("Manifest:", manifest_path)

# Check ZIP archive integrity
for archive in ["/content/dataset.zip",
                "/content/Biomed_SPC Measuring SMR during Active and Resisti/Dataset.zip"]:
    if os.path.isfile(archive):
        with zipfile.ZipFile(archive) as z:
            bad_file = z.testzip()
            print(
                os.path.basename(archive),
                "PASS" if bad_file is None else f"FAIL: {bad_file}"
            )

print("OVERALL:", "PASS" if not failed else "FAIL")
```

```python
%%writefile /content/test_sensor_data_validation.py

import pandas as pd
import numpy as np
import sys

FILE = "/content/Dataset/S01/S01_EEG_01.csv"

EEG_CHANNELS = [
    "F3", "Fz", "FC3", "FC1", "C5", "C3", "CP3", "CP1",
    "P3", "Pz", "PO3", "O1", "P4", "PO4", "Oz", "O2",
    "C4", "C6", "CP2", "CP4", "C1", "Cz", "C2", "CPz",
    "F4", "FCz", "FC2", "FC4"
]

EOG_CHANNELS = ["EOG1", "EOG2", "EOG3", "EOG4"]
ACC_CHANNELS = ["AccX", "AccY", "AccZ"]
REQUIRED = ["Time"] + EEG_CHANNELS + EOG_CHANNELS + ACC_CHANNELS + ["Class"]
VALID_CLASSES = {-1, 0, 1, 2}

def validate_sensor_data(file_path):
    df = pd.read_csv(file_path)
    errors = []

    missing_columns = [c for c in REQUIRED if c not in df.columns]

    if missing_columns:
        errors.append(f"Missing columns: {missing_columns}")
        return errors

    if df.empty:
        errors.append("Dataset contains no records.")
        return errors

    numeric_columns = REQUIRED

    for column in numeric_columns:
        values = pd.to_numeric(df[column], errors="coerce")

        if values.isna().any():
            errors.append(
                f"{column}: {values.isna().sum()} missing or non-numeric values."
            )
        elif not np.isfinite(values).all():
            errors.append(f"{column}: non-finite values detected.")

    if "Class" in df.columns:
        labels = pd.to_numeric(df["Class"], errors="coerce").dropna()
        unexpected = sorted(set(labels.unique()) - VALID_CLASSES)

        if unexpected:
            errors.append(f"Unexpected class labels: {unexpected}")

    print("Rows:", len(df))
    print("Columns:", len(df.columns))
    print("EEG channels:", len(EEG_CHANNELS))
    print("EOG channels:", len(EOG_CHANNELS))
    print("Accelerometer channels:", len(ACC_CHANNELS))
    print("Class labels:", sorted(df["Class"].dropna().unique()))

    return errors


if __name__ == "__main__":
    errors = validate_sensor_data(FILE)

    if errors:
        print("\nFAIL: Sensor data validation found issues.")
        for error in errors:
            print("-", error)
    else:
        print("\nPASS: Sensor data validation successful.")
```

```python
%%writefile /content/test_timestamp_validation.py

import pandas as pd
import numpy as np

FILE = "/content/Dataset/S01/S01_EEG_01.csv"

def validate_timestamps(file_path):
    df = pd.read_csv(file_path)

    if "Time" not in df.columns:
        print("FAIL: Time column is missing.")
        return

    time = pd.to_numeric(df["Time"], errors="coerce")

    missing = int(time.isna().sum())
    non_finite = int((~np.isfinite(time.dropna())).sum())

    valid_time = time.dropna()
    differences = valid_time.diff().dropna()

    duplicates = int(valid_time.duplicated().sum())
    out_of_order = int((differences < 0).sum())
    equal_intervals = int((differences == 0).sum())

    print("Total records:", len(df))
    print("Missing/non-numeric timestamps:", missing)
    print("Non-finite timestamps:", non_finite)
    print("Duplicate timestamps:", duplicates)
    print("Out-of-order intervals:", out_of_order)
    print("Equal consecutive timestamps:", equal_intervals)

    if missing or non_finite or out_of_order:
        print("\nFAIL: Invalid timestamp values or ordering detected.")
    elif duplicates:
        print("\nREVIEW: Timestamps are ordered but contain duplicates.")
    else:
        print("\nPASS: Timestamps are valid and strictly increasing.")


if __name__ == "__main__":
    validate_timestamps(FILE)
```
